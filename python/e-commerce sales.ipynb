```python
#Check for hidden whitespace
categorical_columns = {
    "orders": ["status", "payment_method", "discount_code"],
    "products": ["category"],
    "customers": ["state", "acquisition_channel"]
}

for table_name, columns in categorical_columns.items():
    print(f"\n--- {table_name.upper()} ---")
    
    df = {
        "orders": orders_clean,
        "products": products,
        "customers": customers
    }[table_name]
    
    for col in columns:
        whitespace_count = (
            df[col]
            .dropna()
            .astype(str)
            .str.strip()
            .ne(df[col].dropna().astype(str))
            .sum()
        )
        
        print(f"{col}: {whitespace_count} values with leading/trailing spaces")
```

```python
import pandas as pd
from sqlalchemy import text

# Reload the cleaned CSV
order_items_df = pd.read_csv("order_items_clean.csv")

# Convert missing values to SQL NULL
order_items_df = order_items_df.astype(object).where(
    pd.notna(order_items_df),
    None
)

# Clear the table before importing
with engine.begin() as connection:
    connection.execute(text("DELETE FROM order_items"))

# Import all rows
order_items_df.to_sql(
    name="order_items",
    con=engine,
    if_exists="append",
    index=False,
    chunksize=1000,
    method="multi"
)

print("Order items imported successfully!")
```
